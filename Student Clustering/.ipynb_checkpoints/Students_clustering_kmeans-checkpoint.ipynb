```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.cluster import KMeans
```

```python
students_df = pd.read_csv('student_clustering.csv')
students_df.head()
```

```python
students_df.isna().sum()
```

```python
sns.scatterplot(x=students_df['cgpa'], y=students_df['iq'])
```

```python
wcss = []
for i in range(1,11):
    km = KMeans(n_clusters=i)
    km.fit_predict(students_df)
    wcss.append(km.inertia_)
print(wcss)
```

```python
plt.plot(range(1,11), wcss)
plt.show()
```

```python
X = students_df.iloc[:,:].values
km = KMeans(n_clusters=4)
y = km.fit_predict(students_df)
```

```python
y
```

```python
# X[y==0] # [8.36, 93]
# X[y==0,0] # [8.36]
```

```python
sns.scatterplot(x=X[y==0,0], y=X[y==0,1]) # values of X that falls under cluster 0 - first value, second value
sns.scatterplot(x=X[y==1,0], y=X[y==1,1]) # values of X that falls under cluster 1 - first value, second value
sns.scatterplot(x=X[y==2,0], y=X[y==2,1]) # values of X that falls under cluster 2 - first value, second value
sns.scatterplot(x=X[y==3,0], y=X[y==3,1]) # values of X that falls under cluster 3 - first value, second value
```
