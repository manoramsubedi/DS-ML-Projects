## Splitting data into features and labels

```python
X = heart_df.drop(['target'], axis=1)
Y = heart_df['target']
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=2, stratify=Y)
```

## Loading and Training the model

```python
model = LogisticRegression(max_iter=100)
model.fit(X_train, Y_train)
```

```python
# Accuracy on training data
training_pred = model.predict(X_train)
score = metrics.accuracy_score(training_pred, Y_train)
print("Accuracy of the model on the training data: ", score)
```

```python
# Accuracy on test data
test_pred = model.predict(X_test)
score = metrics.accuracy_score(test_pred, Y_test)
print("Accuracy of the model on the test data: ", score)
```

## Making a predictive system

```python
ind=90
input_data = heart_df.iloc[ind].drop(['target'])
actual_value = heart_df.iloc[ind]['target']

# Convert to numpy array and reshape
input_data_np = np.asarray(input_data)
input_data_reshaped = input_data_np.reshape(1, -1)

prediction = model.predict(input_data_reshaped)
if prediction[0] == 1:
    print("The patient doesn't have a heart disease")
else:
    print("The patient is likely to have heart disease")

print("Actual value:", actual_value)
print("Predicted value:", prediction[0])
```
